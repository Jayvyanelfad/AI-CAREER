```python
# ==========================
# USER INPUT
# ==========================

user_sms = input("Enter SMS Message: ")

# ==========================
# PREPROCESS USER MESSAGE
# ==========================

clean_user = preprocess_text(user_sms)

# Display cleaned text
clean_text = clean_user

# ==========================
# CONVERT TO TF-IDF
# ==========================

user_vector = vectorizer.transform([clean_user])

# ==========================
# PREDICT
# ==========================

result = model.predict(user_vector)
prob = model.predict_proba(user_vector)

# ==========================
# DISPLAY RESULT
# ==========================

print("=" * 50)
print("SMS SPAM DETECTION SYSTEM")
print("=" * 50)

print("\nOriginal SMS")
print(user_sms)

print("\nCleaned SMS")
print(clean_text)

print("\nImportant Keywords")
print(clean_text.split())

print("\nPrediction")
print(result[0].upper())

print("\nPrediction Confidence")
print(round(max(prob[0]) * 100, 2), "%")
```
